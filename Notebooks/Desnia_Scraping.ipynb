## 1. Library

```python
!pip install google-play-scraper

# Import library yang akan digunakan
import pandas as pd
from google_play_scraper import reviews_all, reviews, Sort
```

## 2. Parameter Scraping

```python
# Parameter aplikasi
application_id = "com.ismaker.android.simsimi"
max_data = 20000
size_per_batch = 100
```

```python
# Menampilkan parameter
print(f"ID Aplikasi: {application_id}")
print(f"Maksimum ulasan yang diambil: {max_data}")
print(f"Jumlah ulasan per batch: {size_per_batch}")
```

## 3. Inisialisasi Variabel

```python
# Inisialisasi variabel
review_data = []
current_count = 0
token_lanjutan = None

print("Inisialisasi selesai. Mulai proses pengambilan data...")
```

## 4. Scraping

```python
while current_count < max_data:
    result, token_lanjutan = reviews(
        application_id,
        lang='id',
        country='id',
        sort=Sort.NEWEST,
        count=size_per_batch,
        continuation_token=token_lanjutan if current_count > 0 else None
    )

    for entry in result:
        review_data.append({
            "content": entry["content"],
            "score": entry["score"]
        })
        current_count += 1
        print(f"Berhasil mengambil ulasan ke-{current_count}")

        if current_count >= max_data:
            break

    if not token_lanjutan:
        print("Semua ulasan telah diambil.")
        break
```

## 5. Menyimpan Data CSV

```python
# Menyimpan ke CSV
df_reviews = pd.DataFrame(review_data)
df_reviews.to_csv("dataset (2).csv", index=False, encoding="utf-8")
print(f"\nTotal ulasan berhasil diambil: {len(df_reviews)}")
print("Dataset disimpan dengan nama: dataset (2).csv")
```
